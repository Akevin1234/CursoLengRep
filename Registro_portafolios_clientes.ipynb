Gestion de portafolios financieros

```python
# Listas para almacenar información de clientes
clientes = ["Juan Perez", "Maria Rodriguez", "Carlos Lopez","Ana Gomez"]
montos_invertidos = [10000, 15000, 20000,18000]  # en dólares
rendimientos = [5.5, 6.0,4.0, 7.5]  # en porcentaje
```

```python
print(clientes)
print(montos_invertidos)
print(rendimientos)
```

```python
cliente_1 = "Juan Perez"
cliente_2 = "Maria Rodriguez"
cliente_3 = "Carlos Lopez"
cliente_4= "Ana Gomez"
```

```python
# Definir las variables para cada cliente
cliente_1 = {"nombre": "Juan Perez", "monto": 10000, "rendimiento": 5.5}
cliente_2 = {"nombre": "Maria Rodriguez", "monto": 15000, "rendimiento": 6.0}
cliente_3 = {"nombre": "Carlos Lopez", "monto": 20000, "rendimiento": 4.0}
cliente_4 = {"nombre": "Ana Gomez", "monto": 18000, "rendimiento": 7.5}

# Crear clientes
clientes_tuplas = [cliente_1,cliente_2,cliente_3,cliente_4]

# Imprimir la lista de clientes
print(clientes_tuplas)
```

```python
# Crear el diccionario con los datos de los clientes
registro_clientes = {
    "Juan Perez": {"Monto": 10000, "Rendimiento": 5.5},
    "Maria Rodriguez": {"Monto": 15000, "Rendimiento": 6.0},
    "Carlos Lopez": {"Monto": 20000, "Rendimiento": 4.0},
     "Ana Gomez": {"Monto": 18000, "Rendimiento": 7.5}}


# Imprimir el diccionario
print(registro_clientes)
```

```python
#Crear portafolio
portafolio = {"Juan Perez": {"monto_invertido": 10000, "rendimiento":5.5},
    "Maria Rodriguez": {"monto_invertido": 15000, "rendimiento":6.0},
    "Carlos Lopez": {"monto_invertido": 20000, "rendimiento":4.0},
    "Ana Gomez": {"monto_invertido": 18000, "rendimiento":7.5}}
print(type(portafolio))#especificar el tipo
print(portafolio)#imprimir el portafolio
```

```python
for cliente, datos in portafolio.items():
 if datos["rendimiento"] > 5:#Rendmiento superior al 5%
  print(f"{cliente} tiene un rendimiento de {datos['rendimiento']}%")
```

```python
#Mayor rendimiento en dolares
cliente_mayor_rendimiento= None
rendimiento_mayor =0
for cliente, datos in portafolio.items():
 rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento"]) / 100
 if rendimiento_cliente > rendimiento_mayor:
  rendimiento_mayor = rendimiento_cliente
  cliente_mayor_rendimiento = cliente
print(f"El cliente con el mayor rendimiento en dólares es: {cliente_mayor_rendimiento} con un rendimiento de: ${rendimiento_mayor:.2f}")
```

```python
#rendimiento del portafolio
rendimiento_total = 0
for cliente, datos in portafolio.items():
 rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento"]) / 100
 rendimiento_total += rendimiento_cliente

print(f"El rendimiento total del portafolio es: ${rendimiento_total:.2f}")
```

```python
# Seleccionar el cliente
cliente = "Maria Rodriguez"

# Calcular el rendimiento total en dólares
if cliente in registro_clientes:
    monto = registro_clientes[cliente]["Monto"]
    rendimiento_porcentaje = registro_clientes[cliente]["Rendimiento"]
    rendimiento_total = monto * (rendimiento_porcentaje / 100)
    print(f"El rendimiento total de {cliente} es: ${rendimiento_total:.2f}")
else:
    print(f"Cliente {cliente} no se encuentra.")
```

```python
#Calcular lo que invirtio Carlos
portafolio["Carlos Lopez"]["monto_invertido"] = 25000
print("El valor del monto invertido actualizado para Carlos Lopez es $25.000")
```

```python
#Agregar cliente al portafolio
portafolio["Kevin Quintero"] ={"monto_invertido": 90000, "rendimiento": 9.8}
print("Kevin Quintero ha sido agregado al portafolio con un monto invertido de $90.000 y un rendimiento de 9.8%")
```

```python
#Eliminar un cliente del portafolio
del portafolio["Ana Gomez"]
print("Ana Gomez se ha removido del portafolio")
```
